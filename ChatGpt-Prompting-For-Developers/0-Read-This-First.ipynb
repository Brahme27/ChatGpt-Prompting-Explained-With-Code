## 1. `1-Guidelines-For-Prompting.ipynb`

This notebook is your starting point. It teaches the two most important rules for talking to an AI to get what you want. Think of it as learning how to ask questions clearly.

### Main Idea

To get good answers from an AI, you need to:

1. **Write clear and specific instructions.**
2. **Give the model time to "think"** before it answers.

---

### Key Concepts and Code Explained

#### **Setup and Helper Function**

* Imports the `openai` library and loads your secret API key (like a password).
* Defines the `get_completion` helper function to send prompts and get responses.
* Uses `temperature=0` to make AI predictable and precise.

---

### Principle 1: Write Clear Instructions

**Tactic 1: Use Delimiters**

* Use `` ` ``, `"""`, or `< >` to separate instructions from text.
* Prevents confusion about what text to act on.

**Tactic 2: Ask for Structured Output**

* Request specific formats (e.g., JSON, HTML).
* Great for programming since structured data can be reused easily.

**Tactic 3: Check if Conditions are Met**

* Give "if-then-else" style instructions.
* Stops AI from making up unnecessary details.

**Tactic 4: Few-Shot Prompting**

* Provide examples before your real question.
* Helps the AI learn the style, tone, and format you want.

---

### Principle 2: Give the Model Time to "Think"

**Strategy 1: Specify the Steps to Complete a Task**

* Break tasks into ordered steps.
* Helps AI focus on one part at a time.

**Strategy 2: Instruct the Model to Solve First, Then Evaluate**

* Ask AI to solve a problem itself before checking someone else’s answer.
* Prevents bias and ensures logical reasoning.

---

### Model Limitations: Hallucinations

* AI can sometimes "hallucinate" — make up facts that sound real but aren’t.
* Example: Inventing a fake product description.
* Always verify factual outputs.

## 2. `2-iterative-prompt-development.ipynb`

This notebook teaches you that your first prompt is rarely perfect. The key is to **start simple and refine your prompt** until you get exactly what you need.

### Main Idea

Prompting is a cycle:

1. Write a prompt.
2. See the result.
3. Identify issues.
4. Refine the prompt.
5. Repeat until perfect.

---

### Key Concepts and Example

**Initial Prompt**

* Simple: "Write a product description based on these technical specifications."
* Result: Too long.

**Issue 1: Text Too Long**

* Add: `Use at most 50 words.`
* Result: Concise.

**Issue 2: Wrong Focus**

* Add: Audience is "furniture retailers" → focus on materials.
* Result: Highlights cast aluminum shell and base.

**Issue 3: Needs Dimensions in a Table**

* Add: `Include a table of dimensions in HTML format...`
* Result: Perfect HTML output with description and table.

## 7. `7-chatbot.ipynb`

This notebook shows how to build a **chatbot** that maintains context.

### Main Idea

AI has no memory by default → you must send the full conversation history each turn.

---

### Key Concepts and Examples

**1. Chat Format**

* Messages structured as `system`, `user`, or `assistant`.
* `system`: sets persona/behavior.
* `user`: input.
* `assistant`: AI responses.

**2. Maintaining Context**

* Without previous messages → AI forgets.
* With full conversation history → AI remembers.

**3. OrderBot Example**

* A pizza ordering chatbot.
* System prompt defines persona, flow, and menu.
* Interactive UI built with `panel`.
* Logic: collect messages, send full history, update context, and display conversation.
